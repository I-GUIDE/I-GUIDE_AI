```python
from lazypredict.Supervised import LazyClassifier, LazyRegressor
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.utils import shuffle
```

## LazyClassification

This dataset comes from the Diagnostic Wisconsin Breast Cancer Database (https://archive.ics.uci.edu/dataset/17/breast+cancer+wisconsin+diagnostic). A bit more information on the scikit-learn version: https://scikit-learn.org/stable/datasets/toy_dataset.html#breast-cancer-dataset

```python
breast_cancer_df, classification = load_breast_cancer(return_X_y=True, as_frame=True)
breast_cancer_df.head()
```

According to the documentation, we are trying to classify these breast masses into either Malignant [harmful or cancerous] (212 cases) or Benign (357 cases). 

```python
print(f"The classes in the dataset are: {set(classification)}")
print(f"There are {len(classification)} total observations.")
for cla in set(classification):
    print(f"  * There are {len(classification[classification == cla])} data points equal to {cla}")
classification.head()
```

Here we will re-load the data as arrays that are easier for the ML approach to analyze:

```python
data = load_breast_cancer()
X = data.data
y = data.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

You may see an error that says "Click to show javascript error." which can be ignored.

```python
clf = LazyClassifier(verbose=0, ignore_warnings=True, custom_metric=None)
models, predictions = clf.fit(X_train, X_test, y_train, y_test)
```

```python
models
```

## LazyRegression

LazyPredict also works for regression!

This dataset comes from a 2004 paper that analyzed 442 diabetes patients to explore how ten variables (age, sex, BMI, blood pressure, and six blood serum measurements) explain disease progression after one year. https://scikit-learn.org/stable/datasets/toy_dataset.html#diabetes-dataset

```python
diabetes_df, classification = load_diabetes(return_X_y=True, as_frame=True)
diabetes_df.head()
```

```python
diabetes  = load_diabetes()
X, y = shuffle(diabetes.data, diabetes.target, random_state=0)
X = X.astype(np.float32)
```

```python
offset = int(X.shape[0] * 0.9) # save the last 10% for testing
print(offset)
```

```python
X_train, y_train = X[:offset], y[:offset]
X_test, y_test = X[offset:], y[offset:]
```

You may see an error that says "Click to show javascript error." which can be ignored.

```python
y_train
```

```python
reg = LazyRegressor(verbose=0, ignore_warnings=False, custom_metric=None)
models, predictions = reg.fit(X_train, X_test, y_train, y_test)
```

```python
models
```
