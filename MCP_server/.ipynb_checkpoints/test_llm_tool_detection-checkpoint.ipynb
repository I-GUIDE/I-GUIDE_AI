## Cell 1: Environment Setup

# LLM Tool Detection Test

This notebook tests whether an actual LLM (GPT-4) can:
1. **Detect** your MCP tools automatically
2. **Select** the correct tool for a given task
3. **Execute** tools with proper parameters
4. **Chain** multiple tools together

## Setup

Before running:
1. Start MCP server in Terminal: `cd MCP_server && python server.py`
2. Verify server works: Run MCP Inspector in another terminal
3. Run this notebook to test LLM behavior

````python
import os
import sys
from pathlib import Path
from dotenv import load_dotenv

# Setup paths
repo_root = Path().absolute().parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

# Load environment variables
load_dotenv(repo_root / ".env")

print(f"✅ Repo root: {repo_root}")
print(f"✅ Python path configured")
````

## Cell 2: Secure API Key Input

````python
import getpass

# Securely input OpenAI API key
if not os.getenv("OPENAI_API_KEY"):
    os.environ["OPENAI_API_KEY"] = getpass.getpass("Enter OpenAI API Key: ")
    print("✅ API key set")
else:
    print("✅ API key found in environment")
````

## Cell 3: Load MCP Tools

Using the local adapter for reliable tool loading with correct signatures.

````python
from MCP_server.smolagents_adapter import get_smolagents_tools

# Load tools via adapter (preserves correct signatures)
print("📦 Loading MCP tools...")
tools = get_smolagents_tools()

print(f"\n✅ Loaded {len(tools)} tools:")
for i, tool in enumerate(tools, 1):
    tool_name = tool.name if hasattr(tool, 'name') else str(tool)
    print(f"   {i}. {tool_name}")

# Show detailed info for one tool
if tools:
    sample_tool = tools[0]
    print(f"\n📋 Sample Tool Details:")
    print(f"   Name: {sample_tool.name}")
    print(f"   Description: {sample_tool.description[:100]}..." if len(sample_tool.description) > 100 else f"   Description: {sample_tool.description}")
````

## Cell 4: Create Agent

Initialize smolagents with OpenAI model and your MCP tools.

````python
from smolagents import OpenAIServerModel, ToolCallingAgent

# Create OpenAI model instance
print("🤖 Creating agent with GPT-4o-mini...")
model = OpenAIServerModel(
    model_id="gpt-4o-mini",  # Using mini for cost efficiency
    api_key=os.environ["OPENAI_API_KEY"]
)

# Create agent with tools
agent = ToolCallingAgent(
    model=model,
    tools=tools,
    max_steps=5,  # Allow multi-step reasoning
)

print("✅ Agent created with all MCP tools available")
print(f"   Model: gpt-4o-mini")
print(f"   Max steps: 5")
print(f"   Tools available: {len(tools)}")
````

## 🧪 Test Cases

Now we'll test different scenarios to see how the LLM handles tool detection and usage.

### Test 1: Simple Tool Call (Biomass Estimation)

````python
print("="*60)
print("TEST 1: Simple Tool Call")
print("="*60)

query = "What is the estimated biomass for Iowa in 2023?"
print(f"\n📝 Query: {query}\n")

try:
    result = agent.run(query)
    print(f"\n✅ RESULT:\n{result}")
except Exception as e:
    print(f"\n❌ ERROR: {e}")
    import traceback
    traceback.print_exc()
````

**Expected Behavior:**
- LLM should detect `estimate_biomass` tool
- Should extract parameters: `region="Iowa"`, `year=2023`
- Should return mock biomass data

**What to check:**
- Did it choose the correct tool?
- Did it format parameters correctly?
- Did it interpret the response properly?

## 🔍 Advanced: Inspect Tool Schemas

See exactly what the LLM sees when choosing tools.

````python
import json
import inspect

print("🔍 Tool Schemas Sent to LLM:\n")

for i, tool in enumerate(tools[:3], 1):  # Show first 3 tools
    print(f"{i}. {tool.name}")
    print(f"   Description: {tool.description[:150]}..." if len(tool.description) > 150 else f"   Description: {tool.description}")
    
    # Try to show parameters
    if hasattr(tool, 'inputs'):
        print(f"   Parameters: {tool.inputs}")
    elif hasattr(tool, '__wrapped__'):
        sig = inspect.signature(tool.__wrapped__)
        print(f"   Parameters: {list(sig.parameters.keys())}")
    
    print()
````

## 📝 Notes

### Why Local Adapter?

This notebook uses the local `smolagents_adapter` instead of HTTP calls to the MCP server because:
- ✅ **Correct signatures:** Preserves exact function signatures
- ✅ **Better errors:** Python exceptions instead of HTTP 500
- ✅ **Faster:** No HTTP overhead
- ✅ **Reliable:** No signature mismatch issues

You can still verify your MCP server works correctly using MCP Inspector:
```bash
# Terminal 1
cd MCP_server && python server.py

# Terminal 2
npx -y @modelcontextprotocol/inspector
```

### Cost Estimates

Using GPT-4o-mini:
- Input: ~$0.15 per 1M tokens
- Output: ~$0.60 per 1M tokens
- Each test: ~2-5K tokens
- **Total cost for all tests: < $0.05**

### Switching to LangChain

If you decide to migrate from smolagents to LangChain:
1. Replace `get_smolagents_tools()` with `get_langchain_tools()`
2. Replace `ToolCallingAgent` with LangChain agent
3. Tool implementations stay the same!

Your MCP tools are framework-agnostic.
